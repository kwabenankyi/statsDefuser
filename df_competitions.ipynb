```python
import json
from pathlib import Path
import pandas as pd

competitions = []

with open('data/competitions.json', 'r') as comp:
    competitions_json = json.load(comp)
    for item in competitions_json:
        competitions.append(item)

competitions_df = pd.DataFrame(competitions)
```

```python
competition_names = {}
competition_ids = {}

for name in competitions_df['competition_name'].unique():
    comp_id = competitions_df[competitions_df['competition_name'] == name]['competition_id'].values[0]
    competition_ids[comp_id] = name
    competition_names[name] = comp_id
```

```python
import matplotlib.pyplot as plt
# international vs club competitions, male vs female stacked bar chart

competitions_df["competition_international"].value_counts().plot(kind='bar', stacked=True)
plt.title("International vs Club Competitions")
```
